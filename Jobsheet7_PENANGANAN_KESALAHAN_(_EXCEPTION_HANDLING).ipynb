Praktikum 1 : try...except Dasar dalam Metode Kelas

```python
class DataStore:
    def __init__(self, nama_store):
        self._data = {}
        self.nama = nama_store
        print(f"DataStore '{self.nama}' dibuat.")

    def tambah_data(self, key, value):
        """Menambahkan data ke dalam store."""
        self._data[key] = value
        print(" -> Data berhasil ditambahkan.")

    def get_data(self, key):
        """
        Mencoba mendapatkan data berdasarkan key.
        Menangani KeyError jika key tidak ada.
        """
        print(f"[{self.nama}] Mencoba mengambil data dengan key: '{key}'")
        try:
          nilai = self._data[key]
          print(f" -> Data ditemukan: {nilai}")
          return nilai
        except KeyError:
          print(f" -> ERROR: Key '{key}' tidak ditemukan di DataStore'{self.nama}'.")
          return None

# --- Kode Utama ---
if __name__ == "__main__":
    store_produk = DataStore("Toko Produk")
    store_produk.tambah_data("Buku Tulis", 15000)
    store_produk.tambah_data("Pensil 2B", 2000)

    print("\nMengambil data yang ADA:")
    harga_buku = store_produk.get_data("Buku Tulis")
    if harga_buku is not None:
        print(f" Harga Buku Tulis adalah: {harga_buku}")

    print("\nMengambil data yang TIDAK ADA:")
    harga_penggaris = store_produk.get_data("Penggaris")
    if harga_penggaris is None:
        print(" Harga Penggaris tidak ditemukan (sesuai ekspektasi).")
```

Praktikum 2 : Menangani Beberapa Exception Spesifik

```python
class kalkulatorAngka:
    def __init__(self):
        print("Kalkulator Angka siap.")

    def bagi(self, angka1, angka2):
        """
        Membagi angka1 dengan angka2.
        Menangani potensi VAlueError (jika input bukan angka)
        dan ZeroDivisionError (jika pembagi nol) secara terpisah.
        """
        print(f"\nMencoba membagi '{angka1}' dengan '{angka2}'...")
        try:
            num1 = float(angka1)
            num2 = float(angka2)
            hasil = num1 / num2
            print(f" -> HAsil pembagian: {hasil}")
            return hasil
        except ValueError:
            print(" -> ERROR: Input tidak valid! Pastikan kedua input adalah angka.")
            return None
        except ZeroDivisionError:
            print(" -> ERROR: Tidak bisa melakukan pembagian dengan nol.")
            return None
        except Exception as e: # Changed EXception to Exception
            print(f" -> ERROR Lainnya: Terjadi kesalahan - {e}")
            return None

# --- Kode Utama ---
if __name__ == "__main__":
    calc = kalkulatorAngka() # Corrected class instantiation from KalkulatorAngka to kalkulatorAngka

    print("--- Tes Kalkulator ---")
    hasil1 = calc.bagi(100, 5)
    hasil2 = calc.bagi(20, 0)
    hasil3 = calc.bagi("limapuluh", 10)
    hasil4 = calc.bagi(50, "sepuluh")

    print(f"\n--- Tingkasan hasil ---")
    print(f"HAsil 1 (100/5): {hasil1}")
    print(f"HAsil 2 (20/0): {hasil2}")
    print(f"HAsil 3 ('limapuluh'/10): {hasil3}")
    print(f"HAsil 4 (50/'sepuluh'): {hasil4}")
```

Praktikum 3 : Menggunakan else dan finally dalam Metode

```python
import os

class FileProcessor:
    def __init__(self, nama_filee):
        self.nama_file = nama_filee # Corrected: used nama_filee from argument
        self._file_handle = None
        print(f"\n-> FileProcessor dibuat untuk file: '{self.nama_file}'") # Corrected: used self.nama_file

    def baca_dan_proses(self):
        """Membuka, membaca, dan 'memproses' file, dengan cleanup di finally.""" # Closed docstring
        print(f"[{self.nama_file}] Mencoba memproses...") # Corrected indentation
        try:
            print(f" [{self.nama_file}] Membuka file...") # Corrected indentation and typo
            self._file_handle = open(self.nama_file, 'r', encoding='utf-8') # Consistent use of _file_handle
            print(f" [{self.nama_file}] Membaca isi file...")
            isi = self._file_handle.read()
            print(f" -> File berhasil dibaca.")
            panjang_isi = len(isi) # Defined panjang_isi
            print(f" -> Panjang isi: {panjang_isi} karakter.")
        except FileNotFoundError: # Corrected exception name
            print(f" -> ERROR: File '{self.nama_file}' tidak ditemukan.")
        except UnicodeDecodeError: # Corrected exception name
            print(f" -> ERRO: Gagal membaca '{self.nama_file}' sebagai teks (mungkin file biner?).")
        except Exception as e:
            print(f" -> ERROR LAinnya saat memproses file: {type(e).__name__} - {e}")
        else:
            print(f" -> PEMROSESAN BERHASIL: Tidak ada error terjadi pada '{self.nama_file}'.") # Corrected typo
        finally:
            print(f" [{self.nama_file}] Blok Finally: Melakukan cleanup...")
            if self._file_handle and not self._file_handle.closed: # Consistent use of _file_handle
                print(f" Menutup file '{self.nama_file}'...")
                self._file_handle.close() # Consistent use of _file_handle
            else:
                print(f" -> File tidak perlu ditutup (belum dibuka atau sudah ditutup).") # Conditional print
        print(f"[{self.nama_file}] Selesai mencoba proses.")
        print(f"-" * 30)

# --- Kode Utama ---
if __name__ == "__main__":
    NAMA_FILE_SUKSES = "data_penting.txt"
    NAMA_FILE_GAGAL = "file_yang_tidak_ada.log"

    try:
        with open(NAMA_FILE_SUKSES, "w", encoding='utf-8') as f:
            f.write("Ini adalah baris pertama.\n")
            f.write("Baris kedua berisi data penting.\n")
        print(f"File dummy '{NAMA_FILE_SUKSES}' berhasil dibuat.")
    except IOError as e: # Added 'as e' to access error message
        print(f"GAgal membuat file dummy '{NAMA_FILE_SUKSES}': {e}")

    print("\n--- Skenario 1: Proses Sukses ---")
    processor1 = FileProcessor(NAMA_FILE_SUKSES) # Corrected: Missing parenthesis
    processor1.baca_dan_proses()
```

Praktikum 4 : Mengakses Objek Exception (as e)

```python
class KalkulatorAngka:
    def __init__(self):
        print("Kalkulator Angka siap.")

    def bagi (self, angka1, angka2):
        """
        Membagi angka1 dengan angka2
        Menangkap objek exception untuk menampilkan detail error.
        """
        print(f"\nMencoba Membagi '{angka1}' dengan '{angka2}'...")
        try:
            num1 = float(angka1)
            num2 = float(angka2)
            if num2 == 0:
                raise ZeroDivisionError("Pembagian tidak boleh nol secara eksplisit.") # Corrected indentation and typo
            hasil = num1 / num2
            print(f" -> Hasil pembagian: {hasil}")
            return hasil # Corrected typo

        except ValueError as e_val:
            print(f" -> ERROR Konversi: Input tidak valid.")
            print(f" Detail Error (ValueError): {e_val}")
            return None
        except ZeroDivisionError as e_zero:
            print(f" -> ERROR Pembagian: Tidak dapat membagi dengan nol.")
            print(f" Detail Error (ZeroDivisionError): {e_zero}")
            return None
        except Exception as e:
            print(f" -> ERROR Lainnya: Terjadi kesalahan tak terduga.")
            print(f" Detail Error (Lainnya): {type(e).__name__} - {e}")
            return None

# --- Kode Utama ---
if __name__ == "__main__":
    calc = KalkulatorAngka()

    print("--- Tes Kalkulator ---")
    hasil1 = calc.bagi(100, 5)
    hasil2 = calc.bagi(20, 0)
    hasil3 = calc.bagi("limapuluh", 10)
    hasil4 = calc.bagi(10, [])

    print("\n--- Ringkasan Hasil ---") # Corrected typo 'n---' to '\n---'
    print(f"Hasil 1 (100/5): {hasil1}")
    print(f"Hasil 2 (20/0): {hasil2}")
    print(f"Hasil 3 ('limapuluh'/10): {hasil3}")
    print(f"Hasil 4 (10/[]): {hasil4}")
```

Praktikum 5 : Membangkitkan Exception (raise) dalam Metode

```python
import locale
try:
    locale.setlocale(locale.LC_ALL, 'id_ID.UTF-8')
except locale.Error:
    print("Locale id_ID tidak tersedia.")
def format_rupiah(angka):
    try:
        return locale.currency(angka, grouping=True, symbol='Rp ')
    except ValueError: # Changed from NameError to ValueError to catch locale.currency errors
        return f"Rp {angka:,.2f}"

class PengelolaAkun:
    def __init__(self, nomor_akun, saldo_awal=0):
        print(f"\nMencoba mebuat akun {nomor_akun}...")
        if not isinstance(saldo_awal, (int, float)) or saldo_awal < 0:
              raise ValueError(f"Saldo awal ({saldo_awal}) tidak valid. Harus angka non-negatif.")
        self.nomor_akun = nomor_akun
        self._saldo = float(saldo_awal)
        print(f" -> Akun {self.nomor_akun} berhasil dibuat dengan saldo: {format_rupiah(self._saldo)}")

    def get_saldo(self):
        """
        Mengembalikan saldo saat ini."""
        print(f"[Info Akun {self.nomor_akun}] Saldo saat ini: {format_rupiah(self._saldo)}")
        return self._saldo

    def tarik_tunai(self, jumlah):
        """
        Menarik tunai dengan validasi. Me-raise exception jika gagal."""
        print(f"\n[{self.nomor_akun}] mencoba tarik tunai: {format_rupiah(jumlah)}")

        if not isinstance(jumlah, (int, float)) or jumlah <= 0:
              raise ValueError(f"Jumlah penarikan ({jumlah}) tidak valid. Harus angka positif.")

        if jumlah > self._saldo:
              raise RuntimeError(f"Saldo tidak mencukupi! Saldo: {format_rupiah(self._saldo)}, Diminta: {format_rupiah(jumlah)}")

        self._saldo -= jumlah
        print(f" -> Penarikan {format_rupiah(jumlah)} berhasil.")
        print(f" -> Saldo sekarang: {format_rupiah(self._saldo)}")
        return True

# --- Kode Utama ---
if __name__ == "__main__":
    print("\n--- Skenario 1: Saldo Awal tidak Valid ---")
    try:
        akun_gagal = PengelolaAkun("ACC001", -50000)
    except ValueError as e:
        print(f" ->ERROR DITANGKAP saat pembuatan akun:- {e}")

    print("\n--- Skenario 2: Akun Valid dan Transaksi ---")
    try:
        akun_sukses = PengelolaAkun("ACC002", 100000)
        akun_sukses.tarik_tunai(30000)
        print("\n Mencoba tarik jumlah negatif...")
        akun_sukses.tarik_tunai(-100000)

    except ValueError as e:
        print(f" ERROR DITANGKAP (ValueError) : {e}")
    except RuntimeError as e:
        print(f" ERROR DITANGKAP (RuntimeERROR): {e}")

        print(f"\nSaldo akhir akun ACC002: {format_rupiah(akun_sukses.get_saldo())}")
```

Praktikum 6 : Kombinasi Menangkap dan Membangkitkan Exception

```python
class KonfigurasiError(Exception):
    """Exception khusus untuk menandakan error terkait konfigurasi aplikasi."""
    def __init__(self, message, penyebab_asli=None):
        super().__init__(message)
        self.penyebab_asli = penyebab_asli

    def __str__(self):
        if self.penyebab_asli:
            return f"{super().__str__()} (disebabkan oleh: {type(self.penyebab_asli).__name__})"
        return super().__str__()

class PengelolaSetting:
    def __init__(self, file_config):
        self._config = {} # Initialized _config here for consistency
        self.file_config = file_config
        try:
            self._load_config()
            print(f" -> PengelolaSetting berhasil dibuat untuk '{self.file_config}'.")
        except KonfigurasiError as e:
            print(f" -> GAGAL membuat PengelolaSetting: {e}")
            self._config = {}
            print(" -> Menggunakan konfigurasi default/kosong.")

    def _load_config(self):
        """Metode internal untuk load config. Bisa raise KonfigurasiError."""
        try:
            print(f" Membuka file '[{self.file_config}]'...")
            # Simulate different file contents based on filename
            if self.file_config == 'setting.cfg':
                # Skenario 1: File Salah (simulated malformed content, leads to ValueError for port)
                simulated_data = {"host": "192.168.1.1", "port": "abc"}
                print(f" Membaca dan validasi isi '{self.file_config}'...")
            elif self.file_config == 'config.txt':
                # For 'config.txt', provide valid data to make Skenario 3 'Sukses'.
                # This will make Skenario 2 'Data Port Salah' also load successfully.
                simulated_data = {"host": "192.168.1.1", "port": "8080"}
            else:
                # Default case for any other file
                simulated_data = {"host": "unknown_host", "port": "invalid"}

            port_str = simulated_data['port']
            port_int = int(port_str)
            self._config['port'] = port_int
            self._config['host'] = simulated_data.get('host', 'localhost')
            print(" -> Konfigurasi berhasil di-load dan divalidasi.")

        except FileNotFoundError as e_fnf:
            print(f" ERROR Bawaan: File tidak ditemukan.")
            raise KonfigurasiError(f"File konfigurasi penting '{self.file_config}' hilang.", penyebab_asli=e_fnf) from e_fnf
        except ValueError as e_val:
            print(f" ERROR Bawaan: Nilai konfigurasi tidak valid.")
            raise KonfigurasiError(f"Nilai port dalam '{self.file_config}' tidak valid.", penyebab_asli=e_val) from e_val
        except Exception as e_other:
            print(f" ERROR Bawaan: Terjadi kesalahan tak terduga.")
            raise KonfigurasiError(f"Terjadi error tak terduga saat memuat konfigurasi", penyebab_asli=e_other) from e_other

    def get_setting(self, key, default=None):
        """Mendapatkan nilai setting."""
        return self._config.get(key, default)

# --- kode Utama ---
if __name__ == "__main__":
    print("\n--- Skenario 1: File Salah ---")
    settings1 = PengelolaSetting("setting.cfg")
    print(f" Port dari setting1: {settings1.get_setting('port', 'belum terload')}")

    print("\n--- Skenario 2: Data Port Salah ---")
    setting2 = PengelolaSetting("config.txt")
    print(f" Port dari setting2: {setting2.get_setting('port', 'Belum terload')}")

    print("\n--- Skenario 3: Sukses ---")
    setting3 = PengelolaSetting("config.txt")
    print(f" Port dari setting3: {setting3.get_setting('port')}")
    print(f" Host dari setting3: {setting3.get_setting('host')}")
```

Penugasan :

```python
class InventarisError(Exception):
    """Base class untuk error pada sistem inventaris"""
    pass

class BarangTidakDitemukanError(InventarisError):
    """Terjadi jika mencari barang yang tidak ada di data"""
    pass

class StokTidakCukupError(InventarisError):
    """Terjadi jika pengambilan barang melebihi stok yang ada"""
    pass

class InputNegatifError(InventarisError):
    """Terjadi jika input angka stok bernilai negatif"""
    pass

# --- Kelas Inventaris ---
class InventarisBarang:
    def __init__(self):
        self.stok_barang = {
            "Laptop" : 10,
            "Mouse" : 25,
            "Keyboard" : 15
        }

    def tampilkan_stok(self):
        print("\n--- Stok Barang Saat Ini ---")
        if not self.stok_barang:
            print("Tidak ada barang dalam inventaris.")
        else:
            for nama, jumlah in self.stok_barang.items():
                print(f"- {nama}: {jumlah}")

    def tambah_stok(self, nama, jumlah):
        if jumlah < 0:
            raise InputNegatifError(f"Gagal! Jumlah '{jumlah}' tidak boleh negatif.")

        if nama in self.stok_barang:
            self.stok_barang[nama] += jumlah

        else:
            self.stok_barang[nama] = jumlah
        print(f"Berhasil: Stok {nama} bertambah {jumlah}. Total: {self.stok_barang[nama]}")

    def ambil_barang(self, nama, jumlah):
        if nama not in self.stok_barang:
            raise BarangTidakDitemukanError(f"Error: Barang '{nama}' tidak terdaftar!")

        if jumlah < 0:
            raise InputNegatifError(f"Gagal! Jumlah Pengambilan tidak boleh negatif.")

        if jumlah > self.stok_barang[nama]:
            raise StokTidakCukupError(f"Error: Stok '{nama}' hanya tersisa {self.stok_barang[nama]}")

        self.stok_barang[nama] -= jumlah
        print(f"Berhasil: Mengambil {jumlah} {nama}. Sisa stok: {self.stok_barang[nama]}")

# --- Kode Utama ---
if __name__ == "__main__":
  gudang = InventarisBarang()

  while True:
    gudang.tampilkan_stok()
    print("\nMenu inventaris:")
    print("1. Tambah Stok")
    print("2. Ambil Barang")
    print("3. Keluar")

    pilihan = input("Pilih menu (1/2/3):")

    if pilihan == '3':
        print("Program selesai.")
        break

    try:
        nama_input = input("Masukkan nama barang: ")
        jumlah_input = int(input("Masukan jumlah: "))

        if pilihan == '1':
            gudang.tambah_stok(nama_input, jumlah_input)
        elif pilihan == '2':
            gudang.ambil_barang(nama_input, jumlah_input)
        else:
            print("Menu tidak valid!")

    except ValueError:
        print("ERROR: Input jumlah harus berupa angka bulat!")
    except BarangTidakDitemukanError as e:
        print(e)
    except StokTidakCukupError as e:
        print(e)
    except InputNegatifError as e:
        print(e)
    except Exception as e:
        print(f"Terjadi kesalahan sistem: {e}")
```
