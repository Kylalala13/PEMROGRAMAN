```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

# 1. Bikin dataset dummy Pima Indian Diabetes (768 data, 9 kolom)
np.random.seed(42)
data_pima = {
    'Pregnancies': np.random.randint(0, 15, 768),
    'Glucose': np.random.randint(70, 200, 768),
    'BloodPressure': np.random.randint(40, 120, 768),
    'SkinThickness': np.random.randint(0, 50, 768),
    'Insulin': np.random.randint(0, 300, 768),
    'BMI': np.random.uniform(18.0, 45.0, 768),
    'DiabetesPedigreeFunction': np.random.uniform(0.08, 2.4, 768),
    'Age': np.random.randint(21, 80, 768),
    'Outcome': np.random.randint(0, 2, 768)
}
df = pd.DataFrame(data_pima)

# 2. Tampilkan 5 baris pertama & Info
print("--- 5 Baris Teratas ---")
print(df.head())
print("\n--- Informasi Dataset ---")
df.info()

# 3. Memisahkan Fitur (X) dan Label (y)
X = df[df.columns[:8]]
y = df['Outcome']

# 4. Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

# 5. Fit model SVC Baseline
clf = SVC()
clf.fit(X_train, y_train)

# 6. Menampilkan Skor Akurasi
print("\nSkor Akurasi Baseline SVM:", clf.score(X_test, y_test))
```

```python
import pandas as pd
import numpy as np
from sklearn.svm import SVR
import matplotlib.pyplot as plt

# 1. Bikin dataset dummy Salary_Data (30 baris)
np.random.seed(42)
years_exp = np.sort(np.random.uniform(1.0, 11.0, 30))
salary = 35000 + (years_exp * 9000) + np.random.normal(0, 5000, 30)
data = pd.DataFrame({'YearsExperience': years_exp, 'Salary': salary})

# 2. Cek Info & Head
print("--- Info Dataset ---")
data.info()
print("\n--- 5 Baris Pertama ---")
print(data.head())

# 3. Memisahkan Atribut & Label
X = data['YearsExperience'].values[:, np.newaxis]
y = data['Salary'].values

# 4. Membuat & Melatih Model SVR (C=1000, gamma=0.05, kernel='rbf')
model = SVR(C=1000, gamma=0.05, kernel='rbf')
model.fit(X, y)

# 5. Visualisasi Hasil
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', label='Data Riil')
plt.plot(X, model.predict(X), color='red', label='Prediksi SVR (Baseline)')
plt.title('SVR Baseline Regression')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix, accuracy_score
from matplotlib.colors import ListedColormap

# 1. Bikin dataset dummy iklan_sosmed (400 baris)
np.random.seed(99)
data_sosmed = {
    'User ID': np.arange(15000000, 15000400),
    'Jenis Kelamin': np.random.choice(['Pria', 'Wanita'], 400),
    'Umur': np.random.randint(18, 60, 400),
    'Perkiraan Gaji': np.random.randint(15000, 150000, 400),
    'Transaksi': np.random.choice([0, 1], 400, p=[0.6, 0.4])
}
dataset = pd.DataFrame(data_sosmed)

# 2. Memisahkan Fitur (Umur, Gaji) & Label
X = dataset.iloc[:, 2:-1]
y = dataset.iloc[:, -1]

# 3. Split Dataset (80:20)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=99)

# 4. Convert ke numpy & Normalisasi Skala
X_train = X_train.values
y_train = y_train.values
X_test = X_test.values
y_test = y_test.values

sc = MinMaxScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)

# 5. Pelatihan Model Gaussian Naïve Bayes
classifier = GaussianNB()
classifier.fit(X_train, y_train)

# 6. Prediksi & Evaluasi
y_pred = classifier.predict(X_test)
cm = confusion_matrix(y_test, y_pred)
acc = accuracy_score(y_test, y_pred)

print(f"Akurasi Baseline Naive Bayes : {acc*100:.2f}%")
print("Confusion Matrix :")
print(cm)

# 7. Visualisasi Hasil Testing
X_set, y_set = X_test, y_test
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 0.5, stop = X_set[:, 0].max() + 0.5, step = 0.01),
                     np.arange(start = X_set[:, 1].min() - 0.5, stop = X_set[:, 1].max() + 0.5, step = 0.01))
plt.contourf(X1, X2, classifier.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1],
                c = ListedColormap(('red', 'green'))(i), label = j)
plt.title('Naïve Bayes (Sampel Testing)')
plt.xlabel('Umur (Scaled)')
plt.ylabel('Perkiraan Gaji (Scaled)')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC, SVR
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_squared_error
import warnings

warnings.filterwarnings('ignore')

print("==========================================")
print(" 1. GRID SEARCH FOR SVM (CLASSIFICATION) ")
print("==========================================")
# Menggunakan data dari Percobaan 1 (X_train & X_test)
scaler_svm = StandardScaler()
X_train_scaled = scaler_svm.fit_transform(X_train)
X_test_scaled = scaler_svm.transform(X_test)

param_grid_svm = {
    'C': [0.1, 1, 10, 100],
    'gamma': [1, 0.1, 0.01, 0.001],
    'kernel': ['rbf', 'linear', 'poly']
}

grid_svm = GridSearchCV(SVC(), param_grid_svm, cv=5, scoring='accuracy', n_jobs=1)
grid_svm.fit(X_train_scaled, y_train)

best_svm = grid_svm.best_estimator_
acc_svm_tuned = best_svm.score(X_test_scaled, y_test)

print("Parameter SVM Terbaik :", grid_svm.best_params_)
print(f"Akurasi SVM Setelah Tuning : {acc_svm_tuned * 100:.2f}%\n")


print("==========================================")
print(" 2. GRID SEARCH FOR SVR (REGRESSION)     ")
print("==========================================")
# Mengambil khusus data dari Percobaan 2 (Salary_Data) agar tidak tertimpa Naive Bayes
X_svr = data['YearsExperience'].values.reshape(-1, 1)
y_svr = data['Salary'].values.ravel()

scaler_X = StandardScaler()
scaler_y = StandardScaler()

X_scaled = scaler_X.fit_transform(X_svr)
y_scaled = scaler_y.fit_transform(y_svr.reshape(-1, 1)).ravel()

param_grid_svr = {
    'C': [0.1, 1, 10, 100],
    'gamma': [0.001, 0.01, 0.1, 1],
    'kernel': ['rbf', 'linear']
}

grid_svr = GridSearchCV(SVR(), param_grid_svr, cv=5, scoring='neg_mean_squared_error', n_jobs=1)
grid_svr.fit(X_scaled, y_scaled)

best_svr = grid_svr.best_estimator_

# Prediksi dan kembalikan skala y ke nilai aslinya
y_svr_pred_scaled = best_svr.predict(X_scaled)
y_svr_pred = scaler_y.inverse_transform(y_svr_pred_scaled.reshape(-1, 1)).ravel()

print("Parameter SVR Terbaik :", grid_svr.best_params_)
print(f"R2 Score SVR  : {r2_score(y_svr, y_svr_pred):.4f}")
print(f"MSE SVR       : {mean_squared_error(y_svr, y_svr_pred):.2f}")

# Visualisasi Hasil Tuning SVR
plt.figure(figsize=(8, 5))
plt.scatter(X_svr, y_svr, color='blue', label='Data Real')
plt.plot(X_svr, y_svr_pred, color='green', linewidth=2, label='Prediksi SVR (Tuned)')
plt.title('SVR Fitting Setelah Grid Search Tuning')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.legend()
plt.show()
```

```python
import numpy as np
from sklearn.model_selection import GridSearchCV
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import StandardScaler

print("==========================================")
print(" 3. HYPERPARAMETER TUNING NAÏVE BAYES     ")
print("==========================================")

# Standarisasi fitur khusus untuk Naive Bayes (mencegah bias kelas mayoritas)
scaler_nb = StandardScaler()
X_train_nb = scaler_nb.fit_transform(X_train)
X_test_nb = scaler_nb.transform(X_test)

# Parameter var_smoothing yang lebih aman & responsif
param_grid_nb = {
    'var_smoothing': np.logspace(-9, -1, num=50)
}

grid_nb = GridSearchCV(GaussianNB(), param_grid_nb, cv=5, scoring='accuracy', n_jobs=1)
grid_nb.fit(X_train_nb, y_train)

best_nb = grid_nb.best_estimator_
y_pred_nb_tuned = best_nb.predict(X_test_nb)

# Hitung akurasi baseline pada data scaled
nb_baseline = GaussianNB().fit(X_train_nb, y_train)
acc_baseline = nb_baseline.score(X_test_nb, y_test)
acc_nb_tuned = accuracy_score(y_test, y_pred_nb_tuned)

print("Parameter Naïve Bayes Terbaik :", grid_nb.best_params_)
print(f"Akurasi Naïve Bayes Baseline : {acc_baseline*100:.2f}%")
print(f"Akurasi Naïve Bayes Tuned    : {acc_nb_tuned*100:.2f}%\n")

print("--- Laporan Klasifikasi ---")
print(classification_report(y_test, y_pred_nb_tuned))
```
